```python
import numpy as np
import pandas as pd

df = pd.read_excel("P:/IndusTech/Pune_MIDC_Companies_FINAL.xlsx")
```

```python
df
```

```python
df["valid_phone_no"] = df["phone"].isnull()
```

```python
df["valid_phone_no"] = df["valid_phone_no"].replace({True: "No", False: "Yes"})
```

```python
# Drop the phone column and save the cleaned DataFrame to Excel
df.drop(columns=["phone"], inplace=True)
df.to_excel("P:/IndusTech/Pune_MIDC_Companies_FINAL_valid_phone.xlsx", index=False)
```

```python
import os

# Save as CSV
out_path = r"P:\IndusTech\Pune_MIDC_Companies_FINAL_valid_phone.csv"

# Make sure the folder exists
os.makedirs(os.path.dirname(out_path), exist_ok=True)

# Save the cleaned DataFrame to CSV
df.to_csv(out_path, index=False)
```
